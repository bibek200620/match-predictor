```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("prem_matches.csv")

result_counts = df["FTR"].value_counts().reindex(["H", "D", "A"]).fillna(0).astype(int)
labels = ["Home Win", "Draw", "Away Win"]
colors = ["#2E7D32", "#616161", "#1565C0"]

plt.figure(figsize=(8, 5))
bars = plt.bar(labels, result_counts.values, color=colors)
plt.title("Premier League Match Results Distribution")
plt.xlabel("Full-Time Result")
plt.ylabel("Number of Matches")
plt.grid(axis="y", linestyle="--", alpha=0.3)

for bar, value in zip(bars, result_counts.values):
    plt.text(bar.get_x() + bar.get_width() / 2, value + 0.5, str(value), ha="center", va="bottom")

plt.tight_layout()
plt.show()

result_counts
```
